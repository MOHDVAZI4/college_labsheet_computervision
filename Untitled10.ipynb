```python
import cv2
import numpy as np
from google.colab import files
from IPython.display import display
from PIL import Image

# Upload image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read image
image = cv2.imread(filename)

if image is None:
    print("Could not load image.")
else:
    print("Image loaded successfully!")
    print("Original shape:", image.shape)

    # Convert to grayscale
    gray = cv2.cvtColor(
        image, cv2.COLOR_BGR2GRAY
    )

    print("Grayscale shape:", gray.shape)

    # Display original image
    original_rgb = cv2.cvtColor(
        image, cv2.COLOR_BGR2RGB
    )
    display(Image.fromarray(original_rgb))

    # Display grayscale image
    display(Image.fromarray(gray))

    # Convert grayscale to 3-channel image
    gray_3channel = cv2.cvtColor(
        gray, cv2.COLOR_GRAY2BGR
    )

    # Compare original and grayscale images
    comparison = np.hstack(
        (image, gray_3channel)
    )

    output_filename = "colour_grayscale_comparison.jpg"
    cv2.imwrite(output_filename, comparison)

    print("Saved:", output_filename)

    # Display comparison correctly
    comparison_rgb = cv2.cvtColor(
        comparison, cv2.COLOR_BGR2RGB
    )
    display(Image.fromarray(comparison_rgb))
```

```python
import cv2
import os
from google.colab import drive

# Mount Google Drive
drive.mount('/content/drive')

# Define folders
input_folder = '/content/drive/MyDrive/input_images'
output_folder = '/content/drive/MyDrive/resized_images'

# Create folders if they do not exist
os.makedirs(input_folder, exist_ok=True)
os.makedirs(output_folder, exist_ok=True)

# Set target width
target_width = 300

# Supported image formats
extensions = ('.jpg', '.jpeg', '.png')

# Get all image filenames
image_files = [
    f for f in os.listdir(input_folder)
    if f.lower().endswith(extensions)
]

if not image_files:
    print("No images found in:", input_folder)
else:
    print("Found", len(image_files), "images.")

    # Process each image
    for filename in image_files:
        img_path = os.path.join(
            input_folder, filename
        )

        img = cv2.imread(img_path)

        if img is None:
            print("Could not read:", filename)
            continue

        # Original dimensions
        original_height, original_width = img.shape[:2]

        # Preserve aspect ratio
        ratio = target_width / original_width
        new_width = target_width
        new_height = int(original_height * ratio)

        # Resize image
        resized_img = cv2.resize(
            img,
            (new_width, new_height),
            interpolation=cv2.INTER_AREA
        )

        # Save resized image
        output_filename = "resized_" + filename
        output_path = os.path.join(
            output_folder, output_filename
        )

        cv2.imwrite(output_path, resized_img)

        print(
            f"Resized {filename}: "
            f"{original_width}x{original_height} "
            f"to {new_width}x{new_height}"
        )

    print("Batch resizing complete!")
```

```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read image
image = cv2.imread(filename)

if image is None:
    print("Could not load image.")
else:
    print("Original image shape:", image.shape)

    # Define ROI coordinates
    x = 50
    y = 50
    w = 200
    h = 150

    # Get image dimensions
    img_height, img_width = image.shape[:2]

    # Ensure ROI stays within image boundaries
    if x >= img_width or y >= img_height:
        print("ROI starting point is outside the image.")
    else:
        w = min(w, img_width - x)
        h = min(h, img_height - y)

        # Crop image
        cropped_image = image[y:y+h, x:x+w]

        print("Cropped image shape:", cropped_image.shape)

        # Draw rectangle around ROI
        image_with_roi = image.copy()

        cv2.rectangle(
            image_with_roi,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )

        # Convert images to RGB
        original_rgb = cv2.cvtColor(
            image_with_roi, cv2.COLOR_BGR2RGB
        )
        cropped_rgb = cv2.cvtColor(
            cropped_image, cv2.COLOR_BGR2RGB
        )

        # Display images
        plt.figure(figsize=(12, 6))

        plt.subplot(1, 2, 1)
        plt.imshow(original_rgb)
        plt.title("Original Image with ROI")
        plt.axis("off")

        plt.subplot(1, 2, 2)
        plt.imshow(cropped_rgb)
        plt.title("Cropped Image")
        plt.axis("off")

        plt.tight_layout()
        plt.show()

        # Save and download cropped image
        output_filename = "cropped_image.jpg"

        cv2.imwrite(output_filename, cropped_image)
        print("Cropped image saved:", output_filename)

        files.download(output_filename)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read image
image = cv2.imread(filename)

if image is None:
    print("Could not load image.")
else:
    print("Original image shape:", image.shape)

    # 1. Split into B, G, R channels
    b, g, r = cv2.split(image)

    print("Blue channel shape:", b.shape)
    print("Green channel shape:", g.shape)
    print("Red channel shape:", r.shape)

    # 2. Create zero array
    zeros = np.zeros(
        image.shape[:2], dtype=np.uint8
    )

    # Create individual channel images
    blue_channel_img = cv2.merge(
        [b, zeros, zeros]
    )

    green_channel_img = cv2.merge(
        [zeros, g, zeros]
    )

    red_channel_img = cv2.merge(
        [zeros, zeros, r]
    )

    # Display individual channels
    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(cv2.cvtColor(
        blue_channel_img, cv2.COLOR_BGR2RGB
    ))
    plt.title("Blue Channel")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(cv2.cvtColor(
        green_channel_img, cv2.COLOR_BGR2RGB
    ))
    plt.title("Green Channel")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(cv2.cvtColor(
        red_channel_img, cv2.COLOR_BGR2RGB
    ))
    plt.title("Red Channel")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

    # 3. Reconstruct image
    reconstructed_image = cv2.merge([b, g, r])

    print(
        "Reconstructed image shape:",
        reconstructed_image.shape
    )

    # Display reconstructed image
    plt.figure(figsize=(6, 6))

    plt.imshow(cv2.cvtColor(
        reconstructed_image, cv2.COLOR_BGR2RGB
    ))

    plt.title("Reconstructed Image")
    plt.axis("off")
    plt.show()

    # 4. Verify pixel-level accuracy
    if np.array_equal(image, reconstructed_image):
        print(
            "Pixel-level accuracy verified: "
            "Images are identical."
        )
    else:
        print("Images are not identical.")

    # Calculate difference
    difference = cv2.subtract(
        image, reconstructed_image
    )

    b_diff, g_diff, r_diff = cv2.split(difference)

    if (
        cv2.countNonZero(b_diff) == 0
        and cv2.countNonZero(g_diff) == 0
        and cv2.countNonZero(r_diff) == 0
    ):
        print(
            "Difference image is all zeros. "
            "Pixel-level accuracy confirmed."
        )
    else:
        print("Difference image contains non-zero pixels.")
```

```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read image
image = cv2.imread(filename)

if image is None:
    print("Could not load image.")
else:
    # Make a copy of the original image
    watermarked_image = image.copy()

    # Define watermark properties
    text = "Sample Watermark"
    font = cv2.FONT_HERSHEY_SIMPLEX
    font_scale = 1
    font_thickness = 2
    text_color = (255, 255, 255)  # White in BGR

    # Calculate text size
    (text_width, text_height), baseline = cv2.getTextSize(
        text, font, font_scale, font_thickness
    )

    # Position watermark at bottom-right
    padding = 20

    x_pos = (
        watermarked_image.shape[1]
        - text_width - padding
    )

    y_pos = watermarked_image.shape[0] - padding

    # Add watermark
    cv2.putText(
        watermarked_image,
        text,
        (x_pos, y_pos),
        font,
        font_scale,
        text_color,
        font_thickness,
        cv2.LINE_AA
    )

    # Display watermarked image
    plt.figure(figsize=(8, 8))

    plt.imshow(cv2.cvtColor(
        watermarked_image, cv2.COLOR_BGR2RGB
    ))

    plt.title("Watermarked Image")
    plt.axis("off")
    plt.show()

    # Save watermarked image
    output_filename = "watermarked_image.jpg"

    cv2.imwrite(output_filename, watermarked_image)

    print("Watermarked image saved:", output_filename)

    # Download output
    files.download(output_filename)
```
